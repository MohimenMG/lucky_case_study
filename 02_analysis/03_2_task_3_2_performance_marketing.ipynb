```python
# Task 3.2 | Performance Marketing
from pathlib import Path

import pandas as pd
from sqlalchemy import create_engine

ROOT = next(
    folder for folder in (Path.cwd(), Path.cwd().parent)
    if (folder / "data" / "cleaned").is_dir()
)
CLEANED = ROOT / "data" / "cleaned"

Q3_START = "2026-07-01"
Q3_END = "2026-10-01"  # Exclusive
EXTRACT_END = "2026-11-06"  # Includes November 5
CAIRO_OFFSET = pd.Timedelta(hours=2)
DECEMBER_BUDGET = 150_000
SHIFT_SHARE = 0.05

SOURCE_TO_CHANNEL = {
    "meta": "Meta", "facebook ads": "Meta", "facebook": "Meta",
    "fb": "Meta", "instagram": "Meta", "ig": "Meta",
    "google": "Google Ads", "google ads": "Google Ads",
    "adwords": "Google Ads", "gads": "Google Ads",
    "tiktok": "TikTok", "tt": "TikTok",
    "organic": "Other", "(direct)": "Other",
    "referral": "Other", "ref": "Other", "invite": "Other",
}
```

```python
# Prepare inputs. Run the Task 2 cleanup notebook first to refresh the CSVs.
spend = pd.read_csv(CLEANED / "marketing_spend.csv")
users = pd.read_csv(CLEANED / "users.csv", parse_dates=["signup_ts", "kyc_verified_ts"])
transactions = pd.read_csv(CLEANED / "transactions.csv", parse_dates=["txn_ts"])

users["utm_mapped"] = users["utm_source_normalized"].map(SOURCE_TO_CHANNEL).fillna("Unknown")
users["signup_cairo"] = users["signup_ts"] + CAIRO_OFFSET
transactions["txn_cairo"] = transactions["txn_ts"] + CAIRO_OFFSET

settings = pd.DataFrame([{
    "q3_start": Q3_START,
    "q3_end": Q3_END,
    "extract_end": EXTRACT_END,
    "december_budget": DECEMBER_BUDGET,
    "shift_share": SHIFT_SHARE,
}])

engine = create_engine("sqlite://")
spend.to_sql("marketing_spend", engine, index=False)
users.to_sql("users", engine, index=False)
transactions.to_sql("transactions", engine, index=False)
settings.to_sql("settings", engine, index=False)

with engine.begin() as connection:
    connection.exec_driver_sql("CREATE INDEX transactions_by_user ON transactions (user_id)")

# The sql extension is provided by JupySQL.
%load_ext sql
%config SqlMagic.displaylimit = 20
%sql engine
```

```python
%%sql
-- Confirm reported CPI using the platform dates as supplied.
WITH q3_spend AS (
    SELECT channel, spend_egp, platform_reported_installs
    FROM marketing_spend CROSS JOIN settings
    WHERE date >= q3_start AND date < q3_end
),
totals AS (
    SELECT channel, SUM(spend_egp) AS spend, SUM(platform_reported_installs) AS installs
    FROM q3_spend
    GROUP BY channel
    UNION ALL
    SELECT 'Blended', SUM(spend_egp), SUM(platform_reported_installs)
    FROM q3_spend
)
SELECT channel, ROUND(spend, 2) AS spend_egp, installs AS reported_installs,
       ROUND(spend / NULLIF(installs, 0), 2) AS reported_cpi_egp
FROM totals
ORDER BY channel = 'Blended', channel;
```

```python
%%sql
DROP VIEW IF EXISTS temp.daily_channel_performance;
DROP VIEW IF EXISTS temp.paid_signup_outcomes;

-- Build one row per paid-channel Q3 signup for funnel and campaign analysis.
-- Use Cairo time for the signup date and UTC elapsed time for the deposit window.
CREATE TEMP VIEW paid_signup_outcomes AS
SELECT u.user_id,
       u.utm_campaign_normalized AS campaign,
       u.governorate,
       DATE(u.signup_cairo) AS signup_day,
       u.utm_mapped AS channel,
       CASE
           WHEN u.kyc_verified_ts < s.extract_end THEN 1
           ELSE 0
       END AS kyc_verified,
       CASE
           WHEN u.kyc_verified_ts < s.extract_end
            AND EXISTS (
                SELECT 1
                FROM transactions AS t
                WHERE t.user_id = u.user_id
                  AND t.status = 'success'
                  AND t.txn_type = 'deposit'
                  AND t.txn_ts >= u.signup_ts
                  AND STRFTIME('%Y-%m-%d %H:%M:%f', t.txn_ts)
                      <= STRFTIME('%Y-%m-%d %H:%M:%f', u.signup_ts, '+7 days')
                  AND t.txn_ts < s.extract_end
            )
           THEN 1
           ELSE 0
       END AS activated
FROM users AS u
CROSS JOIN settings AS s
WHERE u.signup_cairo >= s.q3_start
  AND u.signup_cairo < s.q3_end
  AND u.utm_mapped IN ('Meta', 'Google Ads', 'TikTok');

-- Join attributed signup outcomes to platform spend and reported installs by day and channel.
CREATE TEMP VIEW daily_channel_performance AS
WITH signup_totals AS (
    -- Credit each signup and its eventual activation to its Cairo signup day.
    SELECT signup_day,
           channel,
           COUNT(*) AS signups,
           SUM(activated) AS activations
    FROM paid_signup_outcomes
    GROUP BY signup_day, channel
),
daily_spend AS (
    -- Aggregate Q3 spend and reported installs using platform dates as supplied.
    SELECT m.date AS spend_day,
           m.channel,
           SUM(m.spend_egp) AS spend_egp,
           SUM(m.platform_reported_installs) AS installs
    FROM marketing_spend AS m
    CROSS JOIN settings AS s
    WHERE m.date >= s.q3_start
      AND m.date < s.q3_end
    GROUP BY m.date, m.channel
)
-- Preserve spend days with zero attributed signups or activations.
SELECT d.spend_day,
       d.channel,
       d.spend_egp,
       d.installs,
       COALESCE(u.signups, 0) AS signups,
       COALESCE(u.activations, 0) AS activations
FROM daily_spend AS d
LEFT JOIN signup_totals AS u
    ON u.signup_day = d.spend_day
   AND u.channel = d.channel;

-- Report quarterly funnel volumes and acquisition costs by channel.
SELECT channel,
       SUM(installs) AS reported_installs,
       SUM(signups) AS signups,
       SUM(activations) AS activated_users,
       ROUND(100.0 * SUM(activations) / NULLIF(SUM(signups), 0), 2) AS activation_rate_pct,
       ROUND(SUM(spend_egp) / NULLIF(SUM(installs), 0), 2) AS cpi_egp,
       ROUND(SUM(spend_egp) / NULLIF(SUM(signups), 0), 2) AS cps_egp,
       ROUND(SUM(spend_egp) / NULLIF(SUM(activations), 0), 2) AS cost_per_activation_egp
FROM daily_channel_performance
GROUP BY channel
ORDER BY cost_per_activation_egp;
```

```python
%%sql
-- Compare platform-reported installs with attributed signup outcomes.
WITH platform_installs AS (
    SELECT channel, SUM(installs) AS reported_installs
    FROM daily_channel_performance
    GROUP BY channel
),
signup_funnel AS (
    SELECT channel,
           COUNT(*) AS attributed_signups,
           SUM(kyc_verified) AS verified_users,
           SUM(activated) AS activated_users
    FROM paid_signup_outcomes
    GROUP BY channel
)
SELECT f.channel,
       p.reported_installs,
       f.attributed_signups,
       ROUND(100.0 * f.attributed_signups
             / NULLIF(p.reported_installs, 0), 2) AS signups_per_reported_install_pct,
       f.verified_users,
       ROUND(100.0 * f.verified_users
             / NULLIF(f.attributed_signups, 0), 2) AS signup_kyc_rate_pct,
       f.activated_users,
       ROUND(100.0 * f.activated_users
             / NULLIF(f.attributed_signups, 0), 2) AS signup_activation_rate_pct
FROM signup_funnel AS f
JOIN platform_installs AS p ON p.channel = f.channel
ORDER BY signup_activation_rate_pct DESC;
```

```python
%%sql
-- KYC/address governorate describes signups. Spend is available by day and channel.
SELECT COALESCE(NULLIF(TRIM(governorate), ''), 'Unknown') AS signup_governorate,
       COUNT(*) AS signups, SUM(activated) AS activated_users,
       ROUND(100.0 * AVG(activated), 2) AS activation_rate_pct
FROM paid_signup_outcomes
WHERE channel = 'Meta'
GROUP BY signup_governorate
ORDER BY signups DESC, signup_governorate;
```

```python
%%sql
-- Rank days by spend within each channel. Compare the bottom and top quarters.
WITH spend_quartiles AS (
    SELECT *,
           NTILE(4) OVER (PARTITION BY channel ORDER BY spend_egp, spend_day) AS quartile
    FROM daily_channel_performance
),
outer_quartiles AS (
    SELECT * FROM spend_quartiles WHERE quartile IN (1, 4)
),
daily_metrics AS (
    SELECT channel, quartile, 'Spend' AS metric, spend_egp AS value FROM outer_quartiles
    UNION ALL
    SELECT channel, quartile, 'CPI', spend_egp / NULLIF(installs, 0) FROM outer_quartiles
    UNION ALL
    SELECT channel, quartile, 'CPS', spend_egp / NULLIF(signups, 0) FROM outer_quartiles
    UNION ALL
    SELECT channel, quartile, 'Activation cost', spend_egp / NULLIF(activations, 0) FROM outer_quartiles
),
ranked AS (
    SELECT *,
           ROW_NUMBER() OVER (PARTITION BY channel, quartile, metric ORDER BY value) AS position,
           COUNT(*) OVER (PARTITION BY channel, quartile, metric) AS valid_days
    FROM daily_metrics
    WHERE value IS NOT NULL
),
medians AS (
    -- Average the two middle values for even groups.
    SELECT channel, quartile, metric, valid_days, AVG(value) AS median_egp
    FROM ranked
    WHERE position IN ((valid_days + 1) / 2, (valid_days + 2) / 2)
    GROUP BY channel, quartile, metric, valid_days
)
SELECT channel, metric,
       COALESCE(MAX(CASE WHEN quartile = 1 THEN valid_days END), 0) AS low_valid_days,
       ROUND(MAX(CASE WHEN quartile = 1 THEN median_egp END), 2) AS low_median_egp,
       COALESCE(MAX(CASE WHEN quartile = 4 THEN valid_days END), 0) AS high_valid_days,
       ROUND(MAX(CASE WHEN quartile = 4 THEN median_egp END), 2) AS high_median_egp,
       ROUND(100.0 * (
           MAX(CASE WHEN quartile = 4 THEN median_egp END)
           / NULLIF(MAX(CASE WHEN quartile = 1 THEN median_egp END), 0) - 1
       ), 1) AS change_pct
FROM medians
GROUP BY channel, metric
ORDER BY channel, CASE metric WHEN 'Spend' THEN 1 WHEN 'CPI' THEN 2 WHEN 'CPS' THEN 3 ELSE 4 END;
```

```python
%%sql
-- Compare historical acquisition costs across full Monday–Sunday weeks.
WITH weekly_performance AS (
    -- Assign each spend day to its Monday and sum results by channel and week.
    -- Activations are attributed to the user's signup day in the source view.
    SELECT channel,
           DATE(spend_day, '-6 days', 'weekday 1') AS week_start,
           COUNT(DISTINCT spend_day) AS days_observed,
           SUM(spend_egp) AS spend_egp,
           SUM(installs) AS reported_installs,
           SUM(activations) AS activated_users
    FROM daily_channel_performance
    WHERE channel IN ('Google Ads', 'Meta', 'TikTok')
    GROUP BY channel, DATE(spend_day, '-6 days', 'weekday 1')
)
-- Exclude partial weeks and calculate unit costs from weekly totals.
-- Within each channel, show the highest-spend weeks first.
SELECT channel,
       week_start,
       ROUND(spend_egp, 2) AS spend_egp,
       reported_installs,
       activated_users,
       ROUND(spend_egp / NULLIF(reported_installs, 0), 2) AS cpi_egp,
       ROUND(spend_egp / NULLIF(activated_users, 0), 2) AS cost_per_activation_egp
FROM weekly_performance
WHERE days_observed = 7
ORDER BY channel, spend_egp DESC, week_start;
```

```python
%%sql
-- Sum performance by channel and Monday-start week.
WITH weekly_performance AS (
    SELECT channel,
           DATE(spend_day, '-6 days', 'weekday 1') AS week_start,
           COUNT(DISTINCT spend_day) AS days_observed,
           SUM(spend_egp) AS spend_egp,
           SUM(installs) AS reported_installs,
           SUM(activations) AS activated_users
    FROM daily_channel_performance
    WHERE channel IN ('Google Ads', 'Meta', 'TikTok')
    GROUP BY channel, DATE(spend_day, '-6 days', 'weekday 1')
),

-- Exclude partial weeks and rank the remaining weeks by spend.
ranked_weeks AS (
    SELECT channel,
           1.0 * spend_egp / NULLIF(reported_installs, 0) AS cpi_egp,
           1.0 * spend_egp / NULLIF(activated_users, 0) AS cost_per_activation_egp,
           ROW_NUMBER() OVER (
               PARTITION BY channel ORDER BY spend_egp, week_start
           ) AS spend_rank,
           COUNT(*) OVER (PARTITION BY channel) AS full_weeks
    FROM weekly_performance
    WHERE days_observed = 7
),

-- Select the three lowest-spend and three highest-spend weeks.
spend_groups AS (
    SELECT channel,
           CASE WHEN spend_rank <= 3 THEN 'low' ELSE 'high' END AS spend_group,
           cpi_egp,
           cost_per_activation_egp
    FROM ranked_weeks
    WHERE full_weeks >= 6
      AND (spend_rank <= 3 OR spend_rank > full_weeks - 3)
),

-- With three values, the median is the sum minus the smallest and largest.
group_medians AS (
    SELECT channel,
           spend_group,
           CASE WHEN COUNT(cpi_egp) = 3 THEN
               SUM(cpi_egp) - MIN(cpi_egp) - MAX(cpi_egp)
           END AS median_cpi,
           CASE WHEN COUNT(cost_per_activation_egp) = 3 THEN
               SUM(cost_per_activation_egp)
               - MIN(cost_per_activation_egp)
               - MAX(cost_per_activation_egp)
           END AS median_cost_per_activation
    FROM spend_groups
    GROUP BY channel, spend_group
)

-- Compare higher-spend weeks against the lower-spend baseline.
SELECT low.channel,
       ROUND(low.median_cpi, 2) AS low_spend_cpi_egp,
       ROUND(high.median_cpi, 2) AS high_spend_cpi_egp,
       ROUND(100.0 * (high.median_cpi - low.median_cpi)
             / NULLIF(low.median_cpi, 0), 2) AS cpi_change_pct,
       ROUND(low.median_cost_per_activation, 2) AS low_spend_activation_cost_egp,
       ROUND(high.median_cost_per_activation, 2) AS high_spend_activation_cost_egp,
       ROUND(100.0 * (high.median_cost_per_activation - low.median_cost_per_activation)
             / NULLIF(low.median_cost_per_activation, 0), 2) AS activation_cost_change_pct
FROM group_medians AS low
JOIN group_medians AS high
    ON high.channel = low.channel
   AND high.spend_group = 'high'
WHERE low.spend_group = 'low'
ORDER BY low.channel;
```
